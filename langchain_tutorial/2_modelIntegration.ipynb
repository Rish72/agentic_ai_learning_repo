```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GEMINI_API_KEY"] = os.getenv("GEMINI_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
### OpenAI method
from langchain.chat_models import init_chat_model

model = init_chat_model("gpt-4o")
response = model.invoke("What is your input size?")
response.content
```

```python
### Gemini method of invoking
from langchain.chat_models import init_chat_model
model = init_chat_model("google_genai:gemini-3.5-flash-lite")
response = model.invoke("What is the name of the model")
response.content[0]['text']
```

```python
### Groq integration

from langchain.chat_models import init_chat_model
model = init_chat_model("groq:qwen/qwen3.8-27b")
response = model.invoke("Why do some parrots speak properly and some cannot?")
response.content
```

```python
for chunk in model.stream("can you write me 50 words describing about yourself?"):
    print(chunk.text)
```

```python
response = model.batch(["What is the model namel","which company as developed and train the model", "what is the max output and input size of the model"])
for response in response:
    print(response)
```
