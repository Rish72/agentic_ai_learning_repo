# Message Prompts

You can pass list of messages to the model by passing an array of messages

Message Types
- System message: tells model how to behave set their personal and provide context
- Human Message: prompt provided by the user to the model
- AI message: The output by the model to the user
- Tool message: output by the tool call

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages=[
    SystemMessage("You are peotry expert"),
    HumanMessage("Write a poem on artifical intelligence")
]

response = model.invoke(messages)
response.content
```

```python
from langchain.messages import SystemMessage, HumanMessage
from langchain.chat_models import init_chat_model

model = init_chat_model("groq:qwen/qwen3.8-27b")
sys_msg = SystemMessage("you are a helpful agent that helps users to get a good prompt for image generation to help them explore their ideas and imaginations to create images ")
human_msg = HumanMessage(
    role="user",
    content= "Provide me a prompt that generates an image of a person leing on the river side bed, originated from the mountains. Coming down hill. The person should be wearing a trekking outfit leing on the grass surrounded by beautiful flowers all around him",
    metadata= "img_123"
)

response = model.invoke([
    sys_msg,
    human_msg
])
response.content
```
