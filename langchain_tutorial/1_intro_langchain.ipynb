### Langchain introduction

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GEMINI_API_KEY"]= os.getenv("GEMINI_API_KEY")
```

```python
from langchain.agents import create_agent

def get_date_time(country:str) -> str:
    """Get the current data, day, time of the country which user asks"""
    return f"Today is 12/11/2026, Thursday and time is 15:21 of {country}."

agent = create_agent(
    model="google_genai:gemini-3.6-flash",
    tools=[get_date_time],
    system_prompt="You are helpful propmt that helps user to find the currrent date, day and time of any country user asks for"
)
agent
```

```python
response = agent.invoke({"messages": [{"role": "user", "content": "what is the date and time of India"}]})
response["messages"][-1].content[0]["text"]
```
